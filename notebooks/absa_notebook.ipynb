```python

%%writefile processResult.py

import json
import pandas as pd

file_path = "Aspect Term Extraction and Polarity Classification.FAST_LCF_ATEPC.result.json"

with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)

# print first 5
for sent in data[:5]:
    print(f"Comment:    {sent['sentence']}")
    print(f"Aspects:    {sent['aspect']}")
    print(f"Sentiment:  {sent['sentiment']}")
    print(f"Confidence: {sent['confidence']}\n")

# export to csv
rows = []
for item in data:
    rows.append({
        'sentence':   item['sentence'],
        'aspects':    item['aspect'],
        'sentiments': item['sentiment'],
        'confidence': item['confidence']
    })

df = pd.DataFrame(rows)
df.to_csv("absa_results.csv", index=False)
print(f"Exported {len(df)} rows to absa_results.csv")
```

```python
# load original
import pandas as pd
reviews = pd.read_csv("ner_results.csv")

# load absa results
absa_df = pd.read_csv("absa_results.csv")

# merge on sentence = translated_text
merged = reviews.head(len(absa_df)).copy()
merged['absa_aspects']    = absa_df['aspects'].values
merged['absa_sentiments'] = absa_df['sentiments'].values
merged['absa_confidence'] = absa_df['confidence'].values

merged.to_csv("absa_results_full.csv", index=False)
print(f"Exported {len(merged)} rows")
```
